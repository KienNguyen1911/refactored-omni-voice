## 📖 Hướng Dẫn Sử Dụng Nhanh

1. **Voice Design Studio**:
   - Chọn các đặc tính giọng mong muốn: Giới tính, độ tuổi, âm vực (pitch), tốc độ nói (speed), cảm xúc (Whisper, Happy, Angry...).
   - Nhập văn bản và bấm **Tạo giọng nói**.

2. **Voice Cloning (Nhân bản giọng)**:
   - Tải lên file âm thanh mẫu (3 - 15 giây, rõ lời, ít tạp âm).
   - Nhập văn bản mô tả đoạn âm thanh mẫu đang nói (Ground-truth prompt transcription).
   - Nhập văn bản mới mà bạn muốn đọc bằng giọng đó -> Bấm **Nhân bản giọng nói**.

3. **Voice Library & Task Queue**:
   - Lưu trữ các giọng đã tạo vào thư viện để gọi lại bất kỳ lúc nào.
   - Quản lý lịch sử các task tạo giọng và tải file WAV chất lượng cao.
